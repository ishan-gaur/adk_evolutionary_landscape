```python
import csv
from pathlib import Path
import pandas as pd
```

```python
# File paths
DATA_FOLDER = Path("~/projects/adk_evolutionary_landscape/data/")
msa_dataset_path = DATA_FOLDER / "adk_msa_temp.csv"
ortho_dataset_path = DATA_FOLDER / "adk_ml_dataset.csv"
```

```python
# Load the main dataset of sequences, organisms, lid-types, and temperatures
msa_dataset = pd.read_csv(msa_dataset_path)
msa_dataset.head()
```

```python
msa_sequence_cols = msa_dataset.to_numpy()[:, 0+2:499+2]
msa_sequences = ["".join(r.tolist()).replace("-", "") for r in msa_sequence_cols]
msa_sequences[:5]
```

```python
msa_dataset["sequence"] = msa_sequences
msa_dataset = msa_dataset.astype({"sequence": "string"})
for i in range(0, 499):
    if str(i) in msa_dataset.columns:
        del msa_dataset[str(i)]
print(msa_dataset.dtypes)
msa_dataset.head()
```

```python
# Load the ortholog dataset and add to the corresponding rows in the msa_dataset
ortho_dataset = pd.read_csv(ortho_dataset_path)
ortho_dataset = ortho_dataset.astype({"sequence": "string"})
print(ortho_dataset.dtypes)
ortho_dataset.head()
```

```python
# Turns out there are duplicates that seem okay, the only difference is 1 degree in temperature 
for sequence in ortho_dataset["sequence"].tolist():
    ct = msa_dataset["sequence"].tolist().count(sequence)
    if ct == 0:
        print("Missing sequence:", sequence)
    elif ct > 1:
        print(f"Duplicate sequence ({ct}):", sequence)
        indices = [i for i, x in enumerate(msa_dataset["sequence"].tolist()) if x == sequence]
        for i in indices:
            print(msa_dataset.iloc[i])
```

```python
# Join on sequence and leave missing ones empty
full_dataset = msa_dataset.join(ortho_dataset.set_index("sequence"), on="sequence", how="left", lsuffix="", rsuffix="_ortho", validate="many_to_one")
for i in range(len(full_dataset)):
    if full_dataset.iloc[i]["lid_type"] != full_dataset.iloc[i]["lid_type_ortho"]:
        print("Mismatch lid type:", full_dataset.iloc[i]["lid_type"], full_dataset.iloc[i]["lid_type_ortho"])
    if full_dataset.iloc[i]["temperature"] != full_dataset.iloc[i]["temperature_ortho"]:
        print("Mismatch temperature:", full_dataset.iloc[i]["temperature"], full_dataset.iloc[i]["temperature_ortho"])
full_dataset = full_dataset.drop(columns=["lid_type_ortho", "temperature_ortho"])
full_dataset.head()
```
